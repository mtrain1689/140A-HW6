### import library

```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import scipy
```

# PART A

```python
data = np.loadtxt("data/hw6_file.txt")
data -= np.mean(np.mean(data))

fig = plt.imshow(data, extent = [0, 12e-9, 0, 12e-9], origin = "lower")
plt.xlabel("Distance along x axis")
plt.ylabel("Distance along y axis")

plt.vlines(ymin = 0, ymax = 1.2e-8, x = 0.5e-8, color = "red")
plt.vlines(ymin = 0, ymax = 1.2e-8, x = 0.54e-8, color = "red")
plt.hlines(xmin = 0, xmax = 1.2e-8, y = 0.05e-8, color = "purple")
plt.hlines(xmin = 0, xmax = 1.2e-8, y = 0.08e-8, color = "purple")


txt = "Scanning tunnel microscopy data of unknown material with reciprocal properties"
plt.figtext(0.5, -0.05, txt, wrap = True, horizontalalignment = "center")
```

From the above image, the smaller atomic lattice constant (a) seems to be approximately 0.5 nm. Length of superlattice is 3a. Both are rhombohedral.

### PART B

```python
fig, ax = plt.subplots(figsize = (12, 7))
fig = plt.imshow(data, extent = [0, 12e-9, 0, 12e-9], origin = "lower")
plt.xlim(left = 0.43e-8)
plt.xlim(right = 0.57e-8)
plt.ylim(bottom = 0.01e-8)
plt.ylim(top = 0.21e-8)
plt.quiver([4.75e-9, 0.1e-9], 1, 1, color = "r", units = "xy")

plt.show()
```

### PART C

```python
recip = np.fft.fftshift(np.fft.fft2(data))

b = 2 * np.pi / 0.5*10e-9
k = 0.5 * 255 * 2 * np.pi/(12*10e-9)
fig = plt.imshow(np.abs(recip), cmap = 'Blues_r', extent = (-k/b, k/b, -k/b, k/b), origin = "lower")
plt.show()
```

### PART D
The faint points about certain points represent the vertices of the superlattice. If one was to connect the dots, they would notice the vertices form a hexagon, which corresponds to the  periodic rhombohedral formation in real space.
